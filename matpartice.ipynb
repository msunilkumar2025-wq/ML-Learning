```python

import matplotlib.pyplot as plt

days = ["Mon", "Tue", "Wed", "Thu", "Fri"]

cities = ["New York", "London", "Delhi", "Tokyo"]

temperatures = [
    [22, 23, 21, 24, 25],   
    [18, 19, 17, 20, 21],   
    [30, 32, 31, 33, 34],   
    [25, 26, 24, 27, 28]    
]

fig, axes = plt.subplots(2, 2, figsize=(10, 8))

count = 0
for i in range(2):
    for j in range(2):
        axes[i][j].plot(days, temperatures[count], marker='o')
        axes[i][j].set_title(cities[count])
        axes[i][j].set_xlabel("Day")
        axes[i][j].set_ylabel("Temperature (°C)")
        axes[i][j].grid(True)
        count += 1

fig.suptitle("Weekly Temperature by City", fontsize=16)
fig.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()
```
