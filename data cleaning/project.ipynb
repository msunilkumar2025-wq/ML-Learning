```python
import pandas as pd
```

```python
df=pd.read_csv("product_reviews_raw.csv")
```

```python
#Mising values
df["category"] = df["category"].fillna("Unknown")
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
df["review_text"] = df["review_text"].fillna("No Review")
df["reviewer_name"] = df["reviewer_name"].fillna("No Name")
df["review_date"] = df["review_date"].fillna("No Date")
df["verified_purchase"] = df["verified_purchase"].fillna("Unknown")
df["helpful_votes"] = df["helpful_votes"].fillna("0")

```

```python
#Duplicated values
df[["reviewer_name","review_date"]].duplicated()

df.drop_duplicates()
```
