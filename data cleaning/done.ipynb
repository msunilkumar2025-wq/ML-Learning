```python
import pandas as pd
```

```python
df=pd.read_csv("product_reviews_raw.csv")
```

```python
df.duplicated().sum()
```

```python
df[df.duplicated(keep=False)]
```

```python
df = df.drop_duplicates()
```

```python
text_columns = [
    "product_name",
    "category",
    "review_text",
    "reviewer_name"
]

for col in text_columns:
    df[col] = df[col].str.strip()
```

```python
df["reviewer_name"] = df["reviewer_name"].fillna("Unknown")
```

```python
df["category"] = df["category"].fillna("Unknown")
```

```python
df["verified_purchase"] = df["verified_purchase"].fillna("Unknown")
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
```

```python
df["helpful_votes"] = df["helpful_votes"].fillna(0)
```

```python
df["review_date"] = pd.to_datetime(
    df["review_date"],
    errors="coerce"
)
```

```python
df["rating"].unique()
```

```python
df["review_date"] = df["review_date"].dt.strftime("%d/%m/%Y")
```

```python
df.dtypes
```

```python
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
df["helpful_votes"] = pd.to_numeric(df["helpful_votes"], errors="coerce")
df["price_usd"] = pd.to_numeric(df["price_usd"], errors="coerce")
```

```python
print(df.shape)
print(df.isnull().sum())
print(df.duplicated().sum())
print(df.dtypes)
```

```python
df["product_name"] = df["product_name"].str.upper()
df["category"] = df["category"].str.upper()
df["review_text"] = df["review_text"].str.upper()
```

```python
df["reviewer_name"] = df["reviewer_name"].str.capitalize()
```
