```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
page_count=1

while True:
    URL=f"https://quotes.toscrape.com/page/{page_count}/"
    res= requests.get(URL)

    soup= BeautifulSoup(res.text,"lxml")
    quotes= soup.select("div.quote")

    if not quotes:
        print("no valid pages anymore..")
        break

    with open(f"scarped_data/quotes{page_count}.html","w") as f:
         f.write(res.text)
         print(f"download data from pages{page_count}")

    page_count=page_count+1
```

```python
with open("scarped_data/quotes1.html","r") as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, "lxml")
all_quotes = soup.select("div.quote")

print(all_quotes)
```

```python
all_quotes = soup.select("div.quote")
life_quotes = []

for q in all_quotes:
    all_tags = []

    for tag in q.select(".tag"):
        all_tags.append(tag.get_text())

    if "life" in all_tags:
        text = q.select_one("span.text").get_text()
        author = q.select_one("small.author").get_text()
        life_quotes.append([text, author])

print(life_quotes)
```
